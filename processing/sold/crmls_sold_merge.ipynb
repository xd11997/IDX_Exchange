```python
from pathlib import Path

sold_dir = Path("../../raw/sold")
csv_files = sorted(sold_dir.glob("*.csv"))

def extract_month(filename):
    name = filename.stem  # CRMLSSold202401
    
    # 取最后6位：202401
    yyyymm = name[-6:]
    
    year = yyyymm[:4]
    month = yyyymm[4:]
    
    return f"{year}-{month}"

def load_concat_sold(files):
    dfs = []
    for f in files:
        df = pd.read_csv(f)
        df.columns = df.columns.str.strip()  # 统一列名空格

        df["source_file"] = f.name
        df["source_month"] = extract_month(f)

        dfs.append(df)
    return pd.concat(dfs, ignore_index=True, sort=False)

sold_all = load_concat_sold(csv_files)

print("文件数量：", len(csv_files))
print("合并后行列：", sold_all.shape)
display(sold_all.head())
```

```python
# 1. 检查并删除完全重复行
dup_count = sold_all.duplicated().sum()
print("完全重复行数量：", dup_count)
sold_all = sold_all.drop_duplicates(ignore_index=True, keep="first")
print("去重后行数：", sold_all.shape[0])

# 2. 缺失值统计
miss = sold_all.isna().sum().to_frame("missing_count")
miss["missing_rate"] = miss["missing_count"] / len(sold_all)
miss = miss.sort_values("missing_rate", ascending=False)
display(miss.head(40))

# 3. 数值列异常值检查
num_cols = sold_all.select_dtypes(include=["number"]).columns.tolist()
print("数值列：", num_cols)

outliers = []
for col in num_cols:
    ser = sold_all[col].dropna()
    if len(ser) < 10:
        continue
    q1 = ser.quantile(0.25)
    q3 = ser.quantile(0.75)
    iqr = q3 - q1
    low = q1 - 1.5 * iqr
    high = q3 + 1.5 * iqr
    out_count = ((ser < low) | (ser > high)).sum()
    outliers.append((col, out_count, out_count / len(sold_all)))
outlier_df = pd.DataFrame(outliers, columns=["column", "outlier_count", "outlier_rate"])
outlier_df = outlier_df.sort_values("outlier_rate", ascending=False)
display(outlier_df.head(40))
```

```python
def analyze_missingness(df, time_col='month', min_month_threshold=3):
    result = []

    total_months = df[time_col].nunique()

    for col in df.columns:
        if col == time_col:
            continue

        missing_rate = df[col].isna().mean()

        # 每个月的非缺失率
        monthly_non_missing = df.groupby(time_col)[col].apply(lambda x: x.notna().mean())

        # 有数据的月份数（>0就算有）
        months_with_data = (monthly_non_missing > 0).sum()

        # 判断是否结构性缺失
        structural_missing = months_with_data <= min_month_threshold

        # 决策逻辑
        if missing_rate == 1:
            decision = 'DROP_100%_MISSING'

        elif structural_missing:
            decision = 'DROP_STRUCTURAL_MISSING'

        elif missing_rate > 0.8:
            decision = 'KEEP_HIGH_MISSING_ADD_FLAG'

        elif missing_rate > 0.6:
            decision = 'KEEP_MODERATE_MISSING'

        else:
            decision = 'KEEP'

        result.append({
            'feature': col,
            'missing_rate': round(missing_rate, 4),
            'months_with_data': int(months_with_data),
            'total_months': int(total_months),
            'structural_missing': structural_missing,
            'decision': decision
        })

    result_df = pd.DataFrame(result).sort_values(by='missing_rate', ascending=False)

    return result_df
```

```python
report = analyze_missingness(sold_all, time_col='source_month')

report.head(20)
```

```python
# 2) 定义“明显错误”规则
current_year = pd.Timestamp.today().year

rules = {}

if "Latitude" in df.columns:
    rules["lat_out_of_range"] = df["Latitude"].notna() & ((df["Latitude"] < -90) | (df["Latitude"] > 90))

if "Longitude" in df.columns:
    rules["lon_out_of_range"] = df["Longitude"].notna() & ((df["Longitude"] < -180) | (df["Longitude"] > 180))

# 价格、面积、数量类：只抓负数这种明显错误
for col in [
    "ParkingTotal", "LotSizeSquareFeet", "LotSizeAcres", "LotSizeArea",
    "StreetNumberNumeric", "DaysOnMarket",
    "BathroomsTotalInteger", "ListPrice", "OriginalListPrice", "ClosePrice",
    "LivingArea", "AssociationFee", "GarageSpaces",
    "BuildingAreaTotal", "BedroomsTotal", "MainLevelBedrooms",
    "BelowGradeFinishedArea", "TaxAnnualAmount", "Stories"
]:
    if col in df.columns:
        rules[f"{col}_negative"] = df[col].notna() & (df[col] < 0)

# 年份类：明显不合理的年份
if "YearBuilt" in df.columns:
    rules["yearbuilt_invalid"] = df["YearBuilt"].notna() & (
        (df["YearBuilt"] < 1800) | (df["YearBuilt"] > current_year + 1)
    )

if "TaxYear" in df.columns:
    rules["taxyear_invalid"] = df["TaxYear"].notna() & (
        (df["TaxYear"] < 1900) | (df["TaxYear"] > current_year + 1)
    )

# 关键ID重复：如果 ListingKey 存在，重复也先标出来
if "ListingKey" in df.columns:
    rules["duplicate_listingkey"] = df["ListingKey"].notna() & df["ListingKey"].duplicated(keep=False)
```

```python
# 3) 生成错误报告
error_report = pd.DataFrame({
    "rule": list(rules.keys()),
    "bad_count": [int(mask.sum()) for mask in rules.values()]
}).sort_values("bad_count", ascending=False)

display(error_report)
```

```python
# 4) 给每一行打标记，看看它命中了哪些规则
bad_mask = pd.Series(False, index=df.index)
bad_reason = pd.Series("", index=df.index, dtype="object")

for rule_name, mask in rules.items():
    bad_mask = bad_mask | mask
    bad_reason = bad_reason.mask(mask, bad_reason.where(~mask, "")).astype(str)

# 更稳一点：收集每行的所有原因
reason_list = []
for i in df.index:
    reasons = [name for name, mask in rules.items() if bool(mask.loc[i])]
    reason_list.append(";".join(reasons))

df["bad_reason"] = reason_list
bad_rows = df[df["bad_reason"] != ""].copy()

display(bad_rows.head(20))
print("bad rows:", bad_rows.shape[0])
```

```python
# 5) 删除这些明显错误行，保留第一版 clean 数据
sold_clean_v1 = df[df["bad_reason"] == ""].copy()

print("原始行数:", df.shape[0])
print("清洗后行数:", sold_clean_v1.shape[0])
print("删掉行数:", df.shape[0] - sold_clean_v1.shape[0])
```

```python
# 6) 导出第一版 CSV
output_path = "../../processed/sold/sold_clean_v1.csv"
sold_clean_v1.drop(columns=["bad_reason"], errors="ignore").to_csv(output_path, index=False)

print("已导出：", output_path)
```
