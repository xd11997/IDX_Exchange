### 1. Check different columns among raw datasets

```python
df_base = pd.read_csv('../../raw/listed/CRMLSListing202401.csv')
base_cols = set(df_base.columns)
```

```python
from pathlib import Path

folder = Path(r"../../raw/listed")
files = sorted(folder.glob("*.csv"))

col_sets = {}

for f in files:
    df = pd.read_csv(f, nrows=5)   # 只读前几行也行，主要是拿列名
    col_sets[f.name] = set(df.columns)

print("baseline:", "CRMLSListing202401.csv")

for fname, cols in col_sets.items():
    missing = base_cols - cols
    extra = cols - base_cols
    print(f"\n{fname}")
    print("missing:", sorted(missing))
    print("extra:", sorted(extra))
```

#### 2.2 Check missing values of the merged dataset

```python
missing_ratio = df.isna().mean().sort_values(ascending=False)
missing_ratio.head(20)
```

```python
# Feature selection: Due to very high mssing ratio, dropped 'TaxYear', 'BelowGradeFinishedArea', 'BusinessType', 'TaxAnnualAmount'
drop_cols = [
    'FireplacesTotal',
    'ElementarySchoolDistrict',
    'CoveredSpaces',
    'MiddleOrJuniorSchoolDistrict',
    'AboveGradeFinishedArea',
    'TaxYear',
    'BelowGradeFinishedArea',
    'BusinessType',
    'TaxAnnualAmount'
]

df = df.drop(columns=drop_cols)
```

### 2.3 Check outliers of the merged dataset

```python
numeric_cols = [
    'ListPrice',
    'LivingArea',
    'BedroomsTotal',
    'BathroomsTotalInteger',
    'LotSizeSquareFeet'
]
```

```python
df[numeric_cols].describe()
```

```python
df.sort_values('ListPrice', ascending=False).head(10)
df[['ListPrice']].sort_values('ListPrice').head(10)
```

```python
import numpy as np

df.loc[df['ListPrice'] <= 0, 'ListPrice'] = np.nan
df.loc[df['LivingArea'] <= 0, 'LivingArea'] = np.nan
df.loc[df['LotSizeSquareFeet'] <= 0, 'LotSizeSquareFeet'] = np.nan
```
